# Why They Return — live on AMD
Gemma 3 12B on this notebook's AMD Radeon GPU (ROCm PyTorch). Paste a product's reviews, get the fit report and the listing fix.

```python
import os
for k in ('HF_HOME', 'HF_HUB_CACHE', 'HUGGINGFACE_HUB_CACHE', 'TRANSFORMERS_CACHE'):
    os.environ[k] = '/root/hf/hub' if k != 'HF_HOME' else '/root/hf'  # weights live outside the 25 GB workspace
import json, torch
from IPython.display import HTML
from wtr.engine import Engine
from wtr.render import to_html
print(torch.cuda.get_device_name(0), '| ROCm', torch.version.hip)
engine = Engine()  # loads Gemma 3 12B once
```

```python
# 30 real reviews of a dress that is in none of the evaluation sets
example = json.load(open('data/live_example.json'))
reviews = example['reviews']
print(len(reviews), 'reviews; first one:\n', reviews[0][:300])
```

```python
result = engine.report(reviews, name='Dress #136110 · 30 of 1,175 reviews')
HTML(to_html(result))
```

```python
print(f"Gemma says runs small: {result['runs_small']:.0%} · renters said, over all {example['n_reviews']} reviews: {example['true_small']:.0%}")
print(f"Gemma says runs large: {result['runs_large']:.0%} · renters said: {example['true_large']:.0%}")
```
